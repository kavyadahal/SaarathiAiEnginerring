```python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer

#Build The dataset : Four tiny reviews
reviews = [
    "the food was great and service was great",
    "terrible food and slow service", 
    "great service , will visit again", 
    "the food was cold and terrible"
]
cv = CountVectorizer()
counts = cv.fit_transform(reviews)
#print(counts)
pd.set_option('display.max_columns', None)
pd.set_option('display.width', 200)
print(cv.get_feature_names_out())
print(pd.DataFrame(counts.toarray() , columns= cv.get_feature_names_out()))
```

# First text classifier

```python
from review_data import make_reviews
from sklearn.feature_extraction.text import TfidfVectorizer , CountVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import make_pipeline
from sklearn.model_selection import cross_val_score

text , labels = make_reviews()

model1 = make_pipeline(TfidfVectorizer() , LogisticRegression(max_iter=2000))
model2 = make_pipeline(CountVectorizer() , LogisticRegression(max_iter=2000))
print(cross_val_score(model1 , text , labels , cv = 5).mean())
print(cross_val_score(model2 , text , labels , cv = 5).mean())

model1.fit(text , labels)
model2.fit(text , labels)
print(model1.predict(["the momo was delicious and the staff were kind"]))
print(model1.predict(["cold food, rude waiter and a long wait"]))    
print(model2.predict(["the momo was delicious and the staff were kind"])) 
print(model2.predict(["the food was not good"]))    

 #Since not is not in the dictonary couldnt give any  relationship

```
