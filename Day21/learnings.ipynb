```python
import numpy as np
```

# One neuron

```python
x = np.array([1.2, 0.8])      # two features
w = np.array([0.9, 0.6])      # one weight per feature (learned)
b = -0.5                     # the bias / intercept (the 'c')
```

```python
#Weighted sum:
z = np.dot(w,x)+b
print(z)
```

```python
# activation (sigmoid squahesh to 0..1)
def sigmoid(v):
    return 1/ (1+np.exp(-v))

out = sigmoid(z) #0.743
print(out , "late" if out >=0.5 else "on time")
```

```python
rng = np.random.default_rng(42); n = 600
x = rng.uniform(-3, 3, n); y = rng.uniform(-3, 3, n)
healthy = ((x**2 + y**2) < 4).astype(int)
X = np.column_stack([x, y])          # ONLY raw x, y
```

```python
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.neural_network import MLPClassifier
net  = make_pipeline(StandardScaler() , MLPClassifier(hidden_layer_sizes=(16,16) , activation='relu' , max_iter=2000) )
```

```python
from sklearn.model_selection import cross_val_score

print(cross_val_score(net  , X , healthy , cv = 5).mean())
```
