```python
import numpy as np
from sklearn.neural_network import MLPClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import cross_val_score
```

```python
rng = np.random.default_rng(42); n = 600
x = rng.uniform(-3, 3, n); y = rng.uniform(-3, 3, n)
healthy = ((x**2 + y**2) < 4).astype(int)
X = np.column_stack([x, y])          # ONLY raw x, y

logistic = make_pipeline(StandardScaler(), LogisticRegression(max_iter=1000))
net = make_pipeline(StandardScaler(),
    MLPClassifier(hidden_layer_sizes=(16, 16), max_iter=2000, random_state=42))
print(cross_val_score(logistic, X, healthy, cv=5).mean())   # 0.695
print(cross_val_score(net,      X, healthy, cv=5).mean())   # 0.98
```
