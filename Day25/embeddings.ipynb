```python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
from review_data import make_reviews          # 60 hand-written reviews
from sklearn.feature_extraction.text import TfidfVectorizer 
from sklearn.decomposition import TruncatedSVD
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import make_pipeline
from sklearn.model_selection import cross_val_score
from sklearn.preprocessing import Normalizer
import numpy as np
```

```python
words  = { #royalty , maleness , womeness
    "king" : np.array([0.9 , 0.9 , 0.1]),
    "queen" : np.array([0.9 , 0.1 , 0.9]),
    "man" : np.array([0.1 , 0.1 , 0.1]),
    "women" : np.array([0.1 , 0.1 , 0.9])
}
def cosine(a,b):
    return np.dot(a,b)/(np.linalg.norm(a) * np.linalg.norm(b))

print(cosine(words["king"] , words["queen"]))
print(cosine(words["king"] , words["women"]))
```

```python
texts , labels = make_reviews()
lsa = make_pipeline(
	TfidfVectorizer(), #wide sparse table
	TruncatedSVD(n_components = 20 , random_state = 0) ,#squeeze to 20 numbers
	Normalizer() #60 reviews * 20 numbers
	)
embeddings = lsa.fit_transform(texts)
```

```python
q = lsa.transform(["the staff were rude ad the food was cold"])[0]
best = np.argsort(-(embeddings@q))[:3]
for i in best: print(texts[i])
```
